```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('iris.csv')

plt.scatter(df['petal_length'], df['petal_width'], color='purple')
plt.title(' Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('iris.csv')
```

```python
import seaborn as sns
df = sns.load_dataset('iris')
sns.boxplot(x='species', y='petal_length', data=df)
plt.title('Petal Length Comparison Between Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
species_count = df['species'].value_counts()
print("1. Flowers belonging to each species:\n", species_count)
print("-" * 30)
```

```python
plt.hist(df['petal_length'])
plt.title('2. Distribplution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```
